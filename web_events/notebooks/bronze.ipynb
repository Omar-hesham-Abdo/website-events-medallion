```python
import requests
import json
import pandas as pd
catalog_name='website'
```

```python
def load_api():
    url='https://skillsbridge-eight.vercel.app/api/analytics/export'
    api_key='api-key'
    response=requests.get(url,headers={'x-api-key':api_key})
    print(response.status_code)
    data=response.json()
# print(data)
    events=(data['data'])
    # print(events)
    return events
data=load_api()
pdf=pd.json_normalize(data)
display(pdf.head())
# print(len(pdf))
```

```python
pdf['metadata'] = pdf['metadata'].apply(json.loads)
metadata_df = pd.json_normalize(pdf['metadata'])
# metadata_df.head()
pdf = pd.concat(
    [pdf.drop(columns=['metadata']), metadata_df],
    axis=1
)
# pdf['skill_level']=pdf['skill_level'].fillna('').astype(str)
# pdf['hash']=pdf['hash'].astype(str)
display(pdf.head(10))
```
